```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned data
data = pd.read_csv("data_cleaned.csv")

# Convert timestamp
data['timestamp'] = pd.to_datetime(data['timestamp'])
```

```python
# PM2.5 Trend
import os

# Create the directory if it doesn't exist
output_dir = "../images/"
os.makedirs(output_dir, exist_ok=True)

plt.figure(figsize=(10,5))
plt.plot(data['timestamp'], data['pm2.5'])
plt.title("PM2.5 Trend Over Time")
plt.xlabel("Time")
plt.ylabel("PM2.5")
plt.savefig(os.path.join(output_dir, "pm25_trend.png"))
plt.show()
```

```python
# Temperature vs PM2.5
plt.figure(figsize=(6,4))
sns.scatterplot(x=data['temperature_2m'], y=data['pm2.5'])
plt.title("Temperature vs PM2.5")
plt.savefig("../images/temperature_vs_pm25.png")
plt.show()
```

```python
# Humidity vs PM2.5
plt.figure(figsize=(6,4))
sns.scatterplot(x=data['relative_humidity_2m'], y=data['pm2.5'])
plt.title("Humidity vs PM2.5")
plt.savefig("../images/humidity_vs_pm25.png")
plt.show()
```

```python
# Wind vs PM2.5
plt.figure(figsize=(6,4))
sns.scatterplot(x=data['wind_speed_10m'], y=data['pm2.5'])
plt.title("Wind Speed vs PM2.5")
plt.savefig("../images/wind_vs_pm25.png")
plt.show()
```

```python
# NDVI vs PM2.5
plt.figure(figsize=(6,4))
sns.scatterplot(x=data['ndvi'], y=data['pm2.5'])
plt.title("NDVI vs PM2.5")
plt.savefig("../images/ndvi_vs_pm25.png")
plt.show()
```
