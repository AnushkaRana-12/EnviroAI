```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# Load cleaned data
data = pd.read_csv("data_cleaned.csv")

# Select features and target
X = data[['temperature_2m', 'relative_humidity_2m', 'wind_speed_10m', 'ndvi']]
y = data['pm2.5']

# Train Random Forest for feature importance
model = RandomForestRegressor()
model.fit(X, y)

# Feature importance
importance = model.feature_importances_
features = ['Temperature', 'Humidity', 'Wind Speed', 'NDVI']
```

```python
import os

# Plot feature importance
plt.figure(figsize=(8,4))
plt.barh(features, importance)
plt.title("Feature Importance")

# Create the directory if it doesn't exist
output_dir = '../images/'
os.makedirs(output_dir, exist_ok=True)

plt.savefig(os.path.join(output_dir, "feature_importance.png"))
plt.show()
```

```python
import os

# Save final dataset
final_data = data[['temperature_2m', 'relative_humidity_2m', 'wind_speed_10m', 'ndvi', 'pm2.5']]

# Create the directory if it doesn't exist
output_dir = '../data/'
os.makedirs(output_dir, exist_ok=True)

final_data.to_csv(os.path.join(output_dir, "final_dataset.csv"), index=False)

print("Feature engineering completed.")
```
